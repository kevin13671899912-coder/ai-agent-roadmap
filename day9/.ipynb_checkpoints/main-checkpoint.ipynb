```python
import json
import operator
from typing import Annotated, Literal, TypedDict

from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph import END, START, StateGraph
from langgraph.types import Command, interrupt


# ==========================================
# 1. 定义 State
# ==========================================
class EmployeeState(TypedDict):
    employee_id: str
    employee: dict | None
    permission_granted: bool
    leader: dict | None
    error: str | None
    logs: Annotated[list[str], operator.add]


# ==========================================
# 2. 准备模拟数据
# ==========================================
EMPLOYEE_DB = {
    "E1001": {
        "id": "E1001",
        "name": "张三",
        "department": "研发部",
        "leader_id": "E9001",
    },
    "E1002": {
        "id": "E1002",
        "name": "李四",
        "department": "设计部",
        "leader_id": "E9001",
    },
    "E9001": {
        "id": "E9001",
        "name": "李经理",
        "department": "研发部",
        "leader_id": None,
    },
}


# ==========================================
# 3. 编写 Node
# ==========================================
def query_employee(state: EmployeeState) -> dict:
    employee_id = state["employee_id"]
    employee = EMPLOYEE_DB.get(employee_id)

    if employee is None:
        return {
            "employee": None,
            "error": f"员工 {employee_id} 不存在",
            "logs": [f"员工查询失败：{employee_id}"],
        }

    return {
        "employee": employee,
        "error": None,
        "logs": [f"员工查询成功：{employee['name']}"],
    }


def check_permission(state: EmployeeState) -> dict:
    employee = state["employee"]

    if employee is None:
        return {
            "permission_granted": False,
            "error": "员工数据不存在，无法检查权限",
            "logs": ["权限检查失败"],
        }

    # 仅 E1001 拥有系统自动授权；其他员工需人工特权审批
    granted = employee["id"] == "E1001"

    return {
        "permission_granted": granted,
        "error": None if granted else "自动权限验证未通过，等待人工特权审批",
        "logs": ["自动权限验证通过" if granted else "自动权限验证未通过"],
    }


def human_approval(state: EmployeeState) -> dict:
    """人工审批节点：触发真正的断点挂起 (interrupt)"""
    approved = interrupt({
        "employee_id": state["employee_id"],
        "message": f"员工 {state['employee_id']} 自动权限未通过，是否赋予人工特权授权？",
    })

    granted = bool(approved)
    return {
        "permission_granted": granted,
        "error": None if granted else "人工特权授权被拒绝",
        "logs": [
            "【人工介入】特权授权通过" if granted else "【人工介入】特权授权拒绝"
        ],
    }


def query_leader(state: EmployeeState) -> dict:
    employee = state["employee"]

    if employee is None:
        return {
            "leader": None,
            "error": "员工信息不存在",
            "logs": ["无法查询领导"],
        }

    leader_id = employee.get("leader_id")

    if leader_id is None:
        return {
            "leader": None,
            "error": None,
            "logs": ["该员工没有直属领导"],
        }

    leader = EMPLOYEE_DB.get(leader_id)

    return {
        "leader": leader,
        "error": None,
        "logs": [
            f"查询到直属领导：{leader['name']}" if leader else "直属领导不存在"
        ],
    }


def employee_not_found(state: EmployeeState) -> dict:
    return {"logs": ["流程结束：员工不存在"]}


def permission_denied(state: EmployeeState) -> dict:
    return {"logs": ["流程结束：权限不足"]}


# ==========================================
# 4. 编写 Router
# ==========================================
def route_after_employee(
    state: EmployeeState,
) -> Literal["check_permission", "employee_not_found"]:
    if state["employee"] is None:
        return "employee_not_found"
    return "check_permission"


def route_after_permission(
    state: EmployeeState,
) -> Literal["query_leader", "human_approval"]:
    if state["permission_granted"]:
        return "query_leader"
    return "human_approval"


def route_after_approval(
    state: EmployeeState,
) -> Literal["query_leader", "permission_denied"]:
    if state["permission_granted"]:
        return "query_leader"
    return "permission_denied"


# ==========================================
# 5. 构建 Graph
# ==========================================
def build_graph():
    builder = StateGraph(EmployeeState)

    # 注册节点
    builder.add_node("query_employee", query_employee)
    builder.add_node("check_permission", check_permission)
    builder.add_node("human_approval", human_approval)
    builder.add_node("query_leader", query_leader)
    builder.add_node("employee_not_found", employee_not_found)
    builder.add_node("permission_denied", permission_denied)

    # 注册边与条件路由
    builder.add_edge(START, "query_employee")

    builder.add_conditional_edges(
        "query_employee",
        route_after_employee,
        {
            "check_permission": "check_permission",
            "employee_not_found": "employee_not_found",
        },
    )

    builder.add_conditional_edges(
        "check_permission",
        route_after_permission,
        {
            "query_leader": "query_leader",
            "human_approval": "human_approval",
        },
    )

    builder.add_conditional_edges(
        "human_approval",
        route_after_approval,
        {
            "query_leader": "query_leader",
            "permission_denied": "permission_denied",
        },
    )

    builder.add_edge("query_leader", END)
    builder.add_edge("employee_not_found", END)
    builder.add_edge("permission_denied", END)

    checkpointer = InMemorySaver()
    return builder.compile(checkpointer=checkpointer)


# ==========================================
# 6. 测试工具函数与 4 大测试场景
# ==========================================
def create_initial_state(employee_id: str) -> EmployeeState:
    return {
        "employee_id": employee_id,
        "employee": None,
        "permission_granted": False,
        "leader": None,
        "error": None,
        "logs": [],
    }


def print_trace_logs(title: str, result_logs: list[str]):
    print(f"\n【{title} Trace Logs】:")
    for idx, log in enumerate(result_logs, 1):
        print(f"  {idx}. {log}")
    print("\n")


# --- 场景 1：员工存在且自动获得权限 ---
def run_scenario_1(graph):
    print("=" * 65)
    print("▶ 场景 1：员工存在且自动有权限 (Target: E1001)")
    print("=" * 65)

    config = {"configurable": {"thread_id": "thread-001"}}
    result = graph.invoke(create_initial_state("E1001"), config=config)

    snapshot = graph.get_state(config)
    print(f"当前待执行节点 snapshot.next: {snapshot.next}")
    print_trace_logs("场景 1", result["logs"])


# --- 场景 2：员工不存在 ---
def run_scenario_2(graph):
    print("=" * 65)
    print("▶ 场景 2：员工不存在 (Target: E9999)")
    print("=" * 65)

    config = {"configurable": {"thread_id": "thread-002"}}
    result = graph.invoke(create_initial_state("E9999"), config=config)

    snapshot = graph.get_state(config)
    print(f"当前待执行节点 snapshot.next: {snapshot.next}")
    print_trace_logs("场景 2", result["logs"])


# --- 场景 3：自动无权限 -> 中断 -> 人工拒绝 (Reject) ---
def run_scenario_3(graph):
    print("=" * 65)
    print("▶ 场景 3：无自动权限 -> interrupt 暂停 -> 人工审批拒绝")
    print("=" * 65)

    config = {"configurable": {"thread_id": "thread-003-reject"}}

    # 步骤 1：触发运行并陷入中断
    graph.invoke(create_initial_state("E1002"), config=config)

    # 步骤 2：读取中断现场
    snapshot = graph.get_state(config)
    print(f"  - 中断后挂起节点 snapshot.next: {snapshot.next}")
    if snapshot.tasks and snapshot.tasks[0].interrupts:
        print(f"  - 捕获中断请求 Payload: {snapshot.tasks[0].interrupts[0].value}")

    # 步骤 3：人工拒绝授权 (Command(resume=False))
    print("  - 外部决策：人工审批【拒绝】(resume=False)，恢复执行...")
    result = graph.invoke(Command(resume=False), config=config)

    snapshot_after = graph.get_state(config)
    print(f"  - 恢复完成后 snapshot.next: {snapshot_after.next}")
    print_trace_logs("场景 3", result["logs"])


# --- 场景 4：自动无权限 -> 中断 -> 人工通过 (Approve) -> 查询领导 ---
def run_scenario_4(graph):
    print("=" * 65)
    print("▶ 场景 4：无自动权限 -> interrupt 暂停 -> 人工审批通过 -> 查到领导")
    print("=" * 65)

    config = {"configurable": {"thread_id": "thread-004-approve"}}

    # 步骤 1：触发运行并陷入中断
    graph.invoke(create_initial_state("E1002"), config=config)

    # 步骤 2：读取中断现场
    snapshot = graph.get_state(config)
    print(f"  - 中断后挂起节点 snapshot.next: {snapshot.next}")
    if snapshot.tasks and snapshot.tasks[0].interrupts:
        print(f"  - 捕获中断请求 Payload: {snapshot.tasks[0].interrupts[0].value}")

    # 步骤 3：人工同意授权 (Command(resume=True))
    print("  - 外部决策：人工审批【授权通过】(resume=True)，恢复执行...")
    resumed_result = graph.invoke(Command(resume=True), config=config)

    snapshot_after = graph.get_state(config)
    print(f"  - 恢复完成后 snapshot.next: {snapshot_after.next}")
    print_trace_logs("场景 4", resumed_result["logs"])


# ==========================================
# 7. 主入口
# ==========================================
if __name__ == "__main__":
    app = build_graph()

    run_scenario_1(app)
    run_scenario_2(app)
    run_scenario_3(app)
    run_scenario_4(app)
```
